```python
import soccerdata as sd

fbref = sd.FBref(leagues="ITA-Serie A", seasons=2024)
print(fbref.__doc__)
```

```python
seriea_schedule = fbref.read_schedule()
seriea_schedule.head()
```

```python
import pandas as pd

# Define stat types and their internal fbref codes
stat_types = {
    "standard": "stats_standard",
    "shooting": "stats_shooting",
    "passing": "stats_passing",
    "passing_types": "stats_passing_types",
    "gca": "stats_gca",
    "defense": "stats_defense",
    "possession": "stats_possession",
    "playing_time": "stats_playing_time",
    "misc": "stats_misc",
    "keepers": "keeper_stats",
    "keepers_adv": "keeper_stats_adv"
}

# Core player identity fields — used for merging across stat types
merge_keys = ["player", "squad", "position", "age", "country", "minutes"]

all_player_match_stats = []

for idx, row in seriea_schedule.iterrows():
    match_id = row["game_id"]
    match_date = row["date"]
    home_team = row["home_team"]
    away_team = row["away_team"]

    stat_dfs = []

    for short_name, fbref_key in stat_types.items():
        try:
            df = fbref.read_player_match_stats(stat_type=fbref_key, match_id=match_id)

            # Get stat columns (everything not in merge keys)
            stat_cols = [col for col in df.columns if col not in merge_keys]

            # Rename stat columns with prefix
            df = df[merge_keys + stat_cols]
            df = df.rename(columns={col: f"{short_name}_{col}" for col in stat_cols})

            stat_dfs.append(df)
        except Exception as e:
            print(f"❌ Failed: {match_id} | {short_name} | Error: {e}")

    if not stat_dfs:
        continue

    # Merge all stat dataframes together for this match
    df_merged = stat_dfs[0]
    for df_next in stat_dfs[1:]:
        df_merged = df_merged.merge(df_next, on=merge_keys, how="outer")

    # Add match metadata
    df_merged["match_id"] = match_id
    df_merged["match_date"] = match_date
    df_merged["home_team"] = home_team
    df_merged["away_team"] = away_team

    all_player_match_stats.append(df_merged)

# Concatenate all player-match rows
df_all_stats = pd.concat(all_player_match_stats, ignore_index=True)

# Sort for time series analysis
df_all_stats = df_all_stats.sort_values(by=["player", "match_date"])

print(f"✅ Final shape: {df_all_stats.shape}")
df_all_stats.head()
```
