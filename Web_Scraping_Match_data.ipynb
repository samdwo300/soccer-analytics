```python
import soccerdata as sd

fbref = sd.FBref(leagues="ITA-Serie A", seasons=2024)
print(fbref.__doc__)
```

```python
seriea_schedule = fbref.read_schedule()
seriea_schedule.head()
```

```python
import pandas as pd

# Define stat types you want to pull
stat_types = ["passing", "shooting", "defense"]

# Define the columns used to uniquely identify a player in a match
merge_keys = ["player", "squad", "position", "age", "country", "minutes"]

# List to hold all player-match rows
all_player_match_stats = []

for idx, row in seriea_schedule.iterrows():
    match_id = row["game_id"]
    match_date = row["date"]
    home_team = row["home_team"]
    away_team = row["away_team"]

    stat_dfs = []

    for stat_type in stat_types:
        try:
            df = fbref.read_player_match_stats(stat_type=stat_type, match_id=match_id)
            
            # Add prefix to stat columns to avoid collisions when merging
            stat_columns = [col for col in df.columns if col not in merge_keys]
            df = df[merge_keys + stat_columns]  # reorder
            df = df.rename(columns={col: f"{stat_type}_{col}" for col in stat_columns})
            
            stat_dfs.append(df)
        except Exception as e:
            print(f"❌ Failed: {match_id} | Stat: {stat_type} | Error: {e}")

    if not stat_dfs:
        continue  # Skip this match if no data was returned

    # Merge all stat DataFrames on player info
    df_merged = stat_dfs[0]
    for df_next in stat_dfs[1:]:
        df_merged = df_merged.merge(df_next, on=merge_keys, how='outer')

    # Add match metadata
    df_merged["match_id"] = match_id
    df_merged["match_date"] = match_date
    df_merged["home_team"] = home_team
    df_merged["away_team"] = away_team

    # Append to full list
    all_player_match_stats.append(df_merged)

# Combine all player-match stats
df_all = pd.concat(all_player_match_stats, ignore_index=True)

# Sort by player and date for time series usage
df_all = df_all.sort_values(by=["player", "match_date"])

print(f"✅ Final shape: {df_all.shape}")
df_all.head()
```
